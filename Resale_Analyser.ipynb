```python
def year_lowest_resale(file_name, town, year):
    f = open(file_name, 'r')
    f.readline()
    
    flats = list(map(lambda line: (int(line.split(',')[0][:4]),
                                     line.split(',')[1],
                                     line.split(',')[2],
                                     float(line.split(',')[3].strip())), f))

    flats = list(filter(lambda x: x[0] == year and x[1] == town, flats))
    
    minimum = flats[0][3]
    for i in range(len(flats)):
        if flats[i][3] < minimum:
            minimum = flats[i][3]

    f.close()
    return(int(minimum))
```

```python
def year_max_resale(lis, town):

    houses = list(filter(lambda x: x[1] == town, lis))
    maximum = houses[0][3]
    for i in range(len(houses)):
        if houses[i][3] > maximum:
            maximum = houses[i][3]

    return maximum

def town_high_low_resale(year, file_name):
    g = open(file_name, 'r')
    g.readline()
    flats = []
    for line in g:
        date, town, street_name, resale_price = line.strip().split(',')
        flats.append((int(date[:4]), town, street_name, float(resale_price)))
        
    flats = list(filter(lambda x: x[0] == year, flats))
    
    towns = []
    for flat in flats:
        if flat[1] not in towns:
            towns.append(flat[1])
    
    max_resale = []
    for town in towns:
        max_resale.append((town, year_max_resale(flats, town)))
    
    minimum, min_town = max_resale[0][1], max_resale[0][0]
    maximum, max_town = max_resale[0][1], max_resale[0][0]
    for i in range(len(max_resale)):
        if max_resale[i][1] > maximum:
            maximum, max_town = max_resale[i][1], max_resale[i][0]
        elif max_resale[i][1] < minimum:
            minimum, min_town = max_resale[i][1], max_resale[i][0]
    g.close()
    return (max_town, min_town)
```
